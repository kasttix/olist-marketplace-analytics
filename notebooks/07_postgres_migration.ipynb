```python
import pandas as pd
from sqlalchemy import create_engine
engine = create_engine('postgresql://myuser:mypassword@127.0.0.1:5432/mydb')
orders_mart = pd.read_csv("../data/processed/order_items_mart.csv")
customers_mart = pd.read_csv("../data/processed/customers_mart.csv")
sellers_mart = pd.read_csv("../data/processed/sellers_mart.csv")
categories_mart = pd.read_csv("../data/processed/categories_mart.csv")

orders_mart.to_sql('orders_mart', engine, if_exists='replace', index=False)
customers_mart.to_sql('customers_mart', engine, if_exists='replace', index=False)
sellers_mart.to_sql('sellers_mart', engine, if_exists='replace', index=False)
categories_mart.to_sql('categories_mart', engine, if_exists='replace', index=False)

print("Все таблицы успешно перенесены в контейнер")
```
